```python
import pandas as pd
import statsmodels.api as sm
```

```python
df = pd.read_excel('D:/Madhubalaji/Probability/Dataset_1.xlsx')
```

```python
Data = df[['PRICE','FLOOR_SIZE', 'NO_OF_BEDROOMS','NO_OF_BATHROOMS', 'PRICE_SQFT']]
```

```python
X = sm.add_constant(df[['PRICE','FLOOR_SIZE', 'NO_OF_BEDROOMS','NO_OF_BATHROOMS', 'PRICE_SQFT']])
Y = Data['PRICE']
```

```python
MODEL = sm.OLS(Y,X).fit()
print(MODEL.summary())
```

```python
p_values = MODEL.pvalues
print(p_values)
```

```python
Confidence_interval = 0.95
```

```python
if(p_values < (1-Confidence_interval)).any():
    print("Reject Null hypothesis in favor of Alternative hypothesis")
else:
    print("Fail to reject Null hypothesis, and there is no significant effect of independent variables on the Price")
```
